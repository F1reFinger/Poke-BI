# Populando o Banco PostgreQSL

Neste notebook é realizada a conexão com o banco conteinizado para criar e popular a tabela de acordo com os dados do arquivo "Complete_Pokedex-Tratada.csv" .

```python
import pandas as pd
import psycopg2

dataFrame = pd.read_csv("Complete_Pokedex-Tratada.csv")

# Conecta no PostgreSQL
conexao = psycopg2.connect(
    host="localhost",
    port=5432,
    database="silver_db",
    user="silver_user",
    password="silver_password"
)

# cria o cursor
cursor = conexao.cursor()

# com o cursos cria a tabela 
cursor.execute("""
CREATE TABLE IF NOT EXISTS pokedex (
    pokedex_number INT NOT NULL PRIMARY KEY,
    pokemon_name VARCHAR(50) NOT NULL,
    type_1 VARCHAR(50) NOT NULL,
    type_2 VARCHAR(50),
    height INT NOT NULL,
    weight INT NOT NULL,
    hit_points INT NOT NULL,
    attack INT NOT NULL,
    defense INT NOT NULL
)
""")

conexao.commit()

# Inserir dados do CSV
with open('Complete_Pokedex-Tratada.csv', 'r') as arqCSV:
    next(arqCSV)  
    cursor.copy_from(arqCSV, 'pokedex', sep=',')

conexao.commit()

# Fechando conexão
cursor.close()
conexao.close()
```
