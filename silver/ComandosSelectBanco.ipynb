# Consultas e Visualizações no Banco

Neste notebook são demonstrados os comandos SELECT para realizar consultas e visualizações no banco de dados que está em um contêiner do Docker.

#### Visualização da tabela pokemon 

```python
import pandas as pd
import psycopg2
import time

while True:
    try:
        conexao = psycopg2.connect(
            host="localhost",
            port=5432,
            database="silver_db",
            user="silver_user",
            password="silver_password"
        )
        break
    except psycopg2.OperationalError:
        print("O banco não está pronto, aguardando 3 segundos...")
        time.sleep(3)

# cria o cursor
cursor = conexao.cursor()

cursor.execute("""
    SELECT * FROM pokemon; 
""")

# visualizar os resultados
dados = cursor.fetchall() # Captura os resultados 
colunas = [desc[0] for desc in cursor.description] # Captura os nomes das colunas
df = pd.DataFrame(dados, columns=colunas) # Cria um DataFrame para exibir
display(df) # Exibe o DataFrame 


# Fechando conexão
cursor.close()
conexao.close()
```

#### Consulta dos Pokémon que são do tipo 'Fire' (Fogo)

#### Top 10 Pokémon Mais Pesados

```python
import pandas as pd
import psycopg2
import time

while True:
    try:
        conexao = psycopg2.connect(
            host="localhost", 
            port=5432, 
            database="silver_db", 
            user="silver_user", 
            password="silver_password"
        )
        break
    except psycopg2.OperationalError:
        print("O banco não está pronto, aguardando 3 segundos...")
        time.sleep(3)

cursor = conexao.cursor()

cursor.execute("""
    SELECT pokemon_name, weight, type_1
    FROM pokemon
    ORDER BY weight DESC
    LIMIT 10;
""")

print("Resultado da consulta:")
dados = cursor.fetchall()
colunas = [desc[0] for desc in cursor.description]
df = pd.DataFrame(dados, columns=colunas)
display(df)

cursor.close()
conexao.close()
print("Conexão fechada.")
```

#### Pokémon Lendários e Míticos da Primeira Geração

```python
import pandas as pd
import psycopg2
import time

while True:
    try:
        conexao = psycopg2.connect(
            host="localhost", 
            port=5432, 
            database="silver_db", 
            user="silver_user", 
            password="silver_password")
        break
    except psycopg2.OperationalError:
        print("O banco não está pronto, aguardando 3 segundos...")
        time.sleep(3)

cursor = conexao.cursor()

cursor.execute("""
    SELECT pokedex_number, pokemon_name, type_1, type_2, total_stats
    FROM pokemon
    WHERE generation = 1 AND (legendary = TRUE OR mythical = TRUE)
    ORDER BY pokedex_number;
""")

print("Resultado da consulta:")
dados = cursor.fetchall()
colunas = [desc[0] for desc in cursor.description]
df = pd.DataFrame(dados, columns=colunas)
display(df)

cursor.close()
conexao.close()
print("Conexão fechada.")
```

#### Pokémon com dois tipos e Total Stats acima de 600

```python
import pandas as pd
import psycopg2
import time

while True:
    try:
        conexao = psycopg2.connect(
            host="localhost", 
            port=5432, 
            database="silver_db", 
            user="silver_user", 
            password="silver_password")
        break
    except psycopg2.OperationalError:
        print("O banco não está pronto, aguardando 3 segundos...")
        time.sleep(3)

cursor = conexao.cursor()

cursor.execute("""
    SELECT pokemon_name, type_1, type_2, total_stats
    FROM pokemon
    WHERE total_stats > 600 AND type_2 != 'Indefinido'
    ORDER BY total_stats DESC;
""")

print("Resultado da consulta:")
dados = cursor.fetchall()
colunas = [desc[0] for desc in cursor.description]
df = pd.DataFrame(dados, columns=colunas)
display(df)

cursor.close()
conexao.close()
print("Conexão fechada.")
```

#### Contagem de Pokémon por Tipo Primário

```python
import pandas as pd
import psycopg2
import time

while True:
    try:
        conexao = psycopg2.connect(
            host="localhost", 
            port=5432, 
            database="silver_db", 
            user="silver_user", 
            password="silver_password")
        break
    except psycopg2.OperationalError:
        print("O banco não está pronto, aguardando 3 segundos...")
        time.sleep(3)

cursor = conexao.cursor()

cursor.execute("""
    SELECT type_1, COUNT(pokedex_number) AS quantidade
    FROM pokemon
    GROUP BY type_1
    ORDER BY quantidade DESC;
""")

print("Resultado da consulta:")
dados = cursor.fetchall()
colunas = [desc[0] for desc in cursor.description]
df = pd.DataFrame(dados, columns=colunas)
display(df)

cursor.close()
conexao.close()
print("Conexão fechada.")
```

#### Consulta: "Canhões de Vidro" (Ataque > 110 e Defesa < 70)

```python
import pandas as pd
import psycopg2
import time

while True:
    try:
        conexao = psycopg2.connect(
            host="localhost", 
            port=5432, 
            database="silver_db", 
            user="silver_user", 
            password="silver_password")
        break
    except psycopg2.OperationalError:
        print("O banco não está pronto, aguardando 3 segundos...")
        time.sleep(3)

cursor = conexao.cursor()

cursor.execute("""
    SELECT pokemon_name, attack, defense, total_stats
    FROM pokemon
    WHERE attack > 110 AND defense < 70
    ORDER BY attack DESC;
""")

print("Resultado da consulta:")
dados = cursor.fetchall()
colunas = [desc[0] for desc in cursor.description]
df = pd.DataFrame(dados, columns=colunas)
display(df)

cursor.close()
conexao.close()
print("Conexão fechada.")
```
